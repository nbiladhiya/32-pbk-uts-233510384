# K-Means Clustering
### Tujuan : mengelompokkan bahan berdasarkan pola harga (murah, sedang, mahal)

```python
df_cluster = df_clean.copy()

df_cluster['harga'] = pd.to_numeric(df_cluster['harga'], errors='coerce')
df_cluster = df_cluster.dropna(subset=['harga'])
```

```python
X = df_cluster[['harga']]
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df_cluster['cluster'] = kmeans.fit_predict(X)
```

```python
df_cluster.head(10)
```

> standarscaler

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster[['harga']])
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)
df_cluster['cluster'] = kmeans.fit_predict(X_scaled)
```

```python
df_cluster.head(10)
```

```python
df_cluster['harga'].describe()
```

> log transform karna data sangat jomplang dari 275 vs 83.000.000

```python
import numpy as np

df_cluster['harga_log'] = np.log1p(df_cluster['harga'])
```

```python
df_cluster['harga_log'].describe()
```

> clustering data baru

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

X = df_cluster[['harga_log']]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_cluster['cluster'] = kmeans.fit_predict(X_scaled)
```

```python
df_cluster['cluster'].value_counts()
```

🟢 Cluster 0 (Mayoritas / murah-menengah)
bahan harga rendah sampai standar
jumlah paling banyak (89 data)

🟡 Cluster 1 (menengah)
bahan dengan harga sedang (60 data)

🔴 Cluster 2 (mahal)
bahan mahal / high value (21 data)

> Plot Cluster

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(df_cluster['nama_bahan'], df_cluster['harga_log'],
            c=df_cluster['cluster'], cmap='viridis')

plt.xticks(rotation=90)
plt.title("Clustering Harga (After Log Transformation)")
plt.show()
```

> Rata - rata tiap cluster

```python
df_cluster.groupby('cluster')['harga'].mean()
```
